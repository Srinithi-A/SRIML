# 1. Ranked Actions + Reason Codes

The ranked action queue prioritises webpages based on historical search performance and content characteristics. Pages with stronger optimisation signals are reviewed first.

| Priority | Action | Reason Code | Description |
|----------|--------|-------------|-------------|
| High | Refresh Content | STALE_CONTENT | Older content that may benefit from updating. |
| High | Improve CTR | LOW_CTR | Page receives impressions but relatively few clicks. |
| Medium | Improve Search Position | LOW_POSITION | Page is close to page one and may benefit from optimisation. |
| Medium | Expand Content | THIN_CONTENT | Content may require additional depth or coverage. |
| Low | Monitor | STABLE_PAGE | No immediate optimisation action required. |

The ranked actions are intended to support decision-making and should always be reviewed before implementation.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Srinithi-A/SRIML/refs/heads/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Example scoring
df["score"] = (
    (df["ctr"] < 0.05).astype(int)
    + (df["avg_position"] > 20).astype(int)
    + (df["content_age_days"] > 365).astype(int)
)

def get_reason(row):
    if row["ctr"] < 0.05:
        return "LOW_CTR"
    elif row["content_age_days"] > 365:
        return "STALE_CONTENT"
    elif row["avg_position"] > 20:
        return "LOW_POSITION"
    else:
        return "STABLE_PAGE"

df["reason_code"] = df.apply(get_reason, axis=1)

action_map = {
    "LOW_CTR": "Improve CTR",
    "STALE_CONTENT": "Refresh Content",
    "LOW_POSITION": "Improve Search Position",
    "STABLE_PAGE": "Monitor"
}

df["action"] = df["reason_code"].map(action_map)

queue = df.sort_values("score", ascending=False)

queue[["score", "reason_code", "action"]].head(20)
```

# 3. Human Review + No-Go List

Before implementing any recommendation, a human reviewer should verify:

- The page is still relevant to current business goals.
- The recommendation aligns with editorial guidelines.
- Recent updates are not already in progress.
- Seasonal or temporary traffic changes are considered.
- No legal or compliance issues are affected.

The following actions should **not** be automated:

- Deleting webpages.
- Publishing rewritten content automatically.
- Changing page titles without editorial approval.
- Removing internal links automatically.
- Making changes to legal, policy, or regulated content.

```python
no_go = [
    "Delete pages automatically",
    "Publish AI-generated changes automatically",
    "Change page titles without review",
    "Modify legal pages",
    "Remove content automatically"
]

print("No-Go List")
for item in no_go:
    print("-", item)
```
