# 3. Leakage Audit

The final feature set was reviewed for possible data leakage.

The following checks were performed:

- Future performance metrics were excluded.
- Label-derived fields were not used as input features.
- Product workflow flags were excluded because they may indirectly contain decision information.
- Client names, URLs, and other identifying information were not used.
- Only historical search and content features available before prediction were included.

No evidence of direct feature leakage was observed in the final feature vector. The selected features represent historical information that would have been available when making the prediction, making the model suitable for decision-support rather than retrospective explanation.

# 4. Claim Rewrite

### Original Claim

### Revised Claim

The model identifies pages that, based on historical search performance and content characteristics, may benefit from further review. The observed relationships are directional and intended to support decision-making rather than prove future search performance or causal improvement.

This wording better reflects the observational nature of the data and follows the careful claim language used throughout the FlyRank research paper. :contentReference[oaicite:2]{index=2}

```python
original_claim = "The model accurately identifies the best pages to optimise."

revised_claim = (
    "The model identifies pages that may benefit from review "
    "based on historical search signals. The results are "
    "observational and intended for decision support rather "
    "than proof of future performance."
)

print("Original Claim:")
print(original_claim)

print("\nRewritten Claim:")
print(revised_claim)
```
