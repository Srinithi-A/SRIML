## Research Question

Can historical SEO signals be used to identify webpages that should be prioritised for content optimisation?

This project supports editorial decision-making by ranking webpages according to their potential need for optimisation. The unit of analysis is an individual webpage, and the output is a ranked recommendation score. Editors can use these recommendations to prioritise content refreshes, improve click-through rates, and optimise search visibility. The recommendations are intended as decision-support rather than automated decisions.

# 3. Methodology

A Decision Tree Classifier was selected because it produces interpretable decision rules while handling nonlinear feature relationships.

Historical search features were used to identify potential optimisation opportunities. A rule-based baseline was developed before model training to provide a transparent comparison.

Validation used the same train-test split for both the baseline and the Decision Tree model. Leakage checks confirmed that future metrics, client identifiers, and label-derived fields were excluded from model training.

```python
feature_columns = [
    "ctr",
    "avg_position",
    "impressions_90d",
    "content_age_days"
]

print("Selected Features")
print(feature_columns)

print("\nDecision Tree selected as the primary model.")
```

# 4. Results (vs Baseline)

The Decision Tree model was compared with the rule-based baseline using the same validation split.

The observed results suggest that the Decision Tree provides more consistent prioritisation while remaining interpretable. The recommendations are intended to support human decision-making rather than predict future search performance.
